```python
import requests

FASTAPI_URL = "https://latepredictor-backend.onrender.com/"
```

### Example of a Prediction Inferencing

```python
url = f"{FASTAPI_URL}/predict"

payload = {
    "datetime_val": "2026-05-06T15:30:00Z",
    "init_latlon": [1.3, 103.8],
    "dest_latlon": [1.35, 103.9],
    "category_id": "3a63ba2d-7b65-4733-b028-29a11f39e560"
}

response = requests.post(url, json=payload)

print(response.status_code)
print(response.json())
```

### Example of a Feedback
